```python
from statsmodels.tsa.stattools import adfuller
```

```python
#Ho: It is non stationary
#H1: It is stationary

def adf_test(series):
    result=adfuller(series)
    print('p- value: {}'.format(result[1]))
    if result[1] <= 0.05:
        print("strong evidence against the null hypothesis, reject the null hypothesis")
    else:
        print("weak evidence against null hypothesis, indicating it is non-stationary ")
```

```python
import numpy as np
from keras.models import Sequential
from sympy import DenseNDimArray
from sklearn.preprocessing import MinMaxScaler

# Assuming model_ARIMA is your fitted ARIMA model

# Step 1: Predict missing portion using ARIMA
# Since ARIMA in statsmodels doesn't directly support out-of-sample forecasting,
# we'll retrain the ARIMA model on the entire dataset (train_data + test_data) and then forecast.
combined_data = np.concatenate((train_data['VWAP'], test_data['VWAP']))
model_ARIMA_full = ARIMA(combined_data, order=(14, 1, 12))
arima_results = model_ARIMA_full.fit()

# Forecast using ARIMA
arima_forecast = arima_results.forecast(steps=len(test_data))

# Step 2: Combine ARIMA forecast with available data
combined_data = np.concatenate((train_data['VWAP'], arima_forecast))

# Step 3: Preprocess data for LSTM
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(combined_data.reshape(-1, 1))

# Function to create LSTM dataset
def create_dataset(data, time_steps):
    X, y = [], []
    for i in range(len(data) - time_steps):
        X.append(data[i:(i + time_steps), 0])
        y.append(data[i + time_steps, 0])
    return np.array(X), np.array(y)

# Define time steps
time_steps = 10  # Adjust based on your requirement

# Create LSTM dataset
X, y = create_dataset(scaled_data, time_steps)

# Reshape data for LSTM [samples, time steps, features]
X = np.reshape(X, (X.shape[0], X.shape[1], 1))

# Step 4: Build LSTM model
model_lstm = Sequential()
model_lstm.add(LSTM(units=50, return_sequences=True, input_shape=(X.shape[1], 1)))
model_lstm.add(LSTM(units=50))
model_lstm.add(Dense(units=1))
model_lstm.compile(optimizer='adam', loss='mean_squared_error')

# Train LSTM model
model_lstm.fit(X, y, epochs=100, batch_size=32)

# Prepare input for prediction
last_time_steps_data = combined_data[-time_steps:]  # Last time steps from combined data
last_scaled_data = scaler.transform(last_time_steps_data.reshape(-1, 1))
last_scaled_data = np.reshape(last_scaled_data, (1, time_steps, 1))

# Predict using LSTM model
lstm_forecast_scaled = model_lstm.predict(last_scaled_data)

# Inverse transform to get actual forecast values
lstm_forecast = scaler.inverse_transform(lstm_forecast_scaled)

# Print forecast
print("LSTM Forecast:", lstm_forecast)
```

```python
import matplotlib.pyplot as plt

# Define the number of additional weeks to forecast
additional_weeks = 2
forecast_steps = additional_weeks * 7  

# Prepare input for additional prediction
additional_forecast = []
last_scaled_data = np.reshape(last_scaled_data, (time_steps, 1))  

for _ in range(forecast_steps):
    # Predict using LSTM model
    lstm_forecast_scaled = model_lstm.predict(np.expand_dims(last_scaled_data, axis=0))
    lstm_forecast = scaler.inverse_transform(lstm_forecast_scaled)
    additional_forecast.append(lstm_forecast[0][0])
    
    # Update last_scaled_data for next prediction
    last_scaled_data = np.roll(last_scaled_data, -1) 
    last_scaled_data[-1] = lstm_forecast_scaled  

# Extend the forecast index
forecast_index_extended = pd.date_range(start=test_data.index[-1], periods=forecast_steps + 1)[1:]  # Skip the first date (already in test_data)

# Plotting the extended LSTM forecast
plt.figure(figsize=(10, 6))
plt.plot(test_data.index, test_data['VWAP'], label='Actual Data')
plt.plot(forecast_index_extended, additional_forecast, marker='o', markersize=8, label='Extended LSTM Forecast')
plt.title('Extended LSTM Forecast')
plt.xlabel('Date')
plt.ylabel('VWAP')
plt.legend()
plt.grid(True)
plt.show()
```
